```python
import pandas as pd

# Read all six CSV files
customers = pd.read_csv("Customer Data.csv")
sellers = pd.read_csv("Sellers.csv")
products = pd.read_csv("Product.csv")
orders = pd.read_csv("Orders.csv")
order_items = pd.read_csv("Order Items.csv")
payments = pd.read_csv("Payments.csv")

# Display the size of each dataset
print("Customers:", customers.shape)
print("Sellers:", sellers.shape)
print("Products:", products.shape)
print("Orders:", orders.shape)
print("Order Items:", order_items.shape)
print("Payments:", payments.shape)

# Check the columns in all six datasets
print("Customers:", customers.columns.tolist())
print("Sellers:", sellers.columns.tolist())
print("Products:", products.columns.tolist())
print("Orders:", orders.columns.tolist())
print("Order Items:", order_items.columns.tolist())
print("Payments:", payments.columns.tolist())
```

```python
import pandas as pd
from sqlalchemy import text

# Disable foreign-key checks while loading the CSV data
with engine.begin() as conn:
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 0"))

# Clear the existing table data
with engine.begin() as conn:
    conn.execute(text("DELETE FROM payments"))
    conn.execute(text("DELETE FROM order_items"))
    conn.execute(text("DELETE FROM orders"))
    conn.execute(text("DELETE FROM product"))
    conn.execute(text("DELETE FROM seller"))
    conn.execute(text("DELETE FROM customers"))

# Import all six datasets
customers.to_sql("customers", con=engine, if_exists="append", index=False)
print("Customers imported:", len(customers))

sellers.to_sql("seller", con=engine, if_exists="append", index=False)
print("Sellers imported:", len(sellers))

# Rename the two columns to match the MySQL table
products_import = products.rename(columns={
    "product_name_lenght": "product_name_length",
    "product_description_lenght": "product_description_length"
})

products_import.to_sql("product", con=engine, if_exists="append", index=False)
print("Products imported:", len(products_import))

orders.to_sql("orders", con=engine, if_exists="append", index=False)
print("Orders imported:", len(orders))

order_items.to_sql("order_items", con=engine, if_exists="append", index=False)
print("Order Items imported:", len(order_items))

payments.to_sql("payments", con=engine, if_exists="append", index=False)
print("Payments imported:", len(payments))

# Re-enable foreign-key checks
with engine.begin() as conn:
    conn.execute(text("SET FOREIGN_KEY_CHECKS = 1"))

print("\nAll six datasets imported successfully.")
```
